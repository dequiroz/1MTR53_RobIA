## 1. Modularidad: Definición de Funciones y Retorno Múltiple

A medida que los algoritmos de control e IA incrementan su extensión, la programación secuencial plana se vuelve ineficiente. La modularidad mediante funciones (`def`) permite encapsular algoritmos reutilizables, reduciendo la duplicación de código y facilitando la depuración.

A diferencia de C/C++, donde la devolución de múltiples variables suele requerir el uso de punteros o referencias pasadas por argumento, Python permite el **retorno múltiple nativo** de datos empaquetando los valores automáticamente en una tupla.

```python
# ==============================================================================
# 1. FUNCIONES Y RETORNO MÚLTIPLE EN CINEMÁTICA
# ==============================================================================
import math

print("--- 1. Cálculo de Transformación Geométrica ---")

def resolver_cinematica_directa(longitud_eslabon, angulo_grados):
    """
    Calcula la posición cartesiana (x, y) del extremo de un eslabón rotacional.

    Parámetros:
        longitud_eslabon (float): Longitud del brazo mecánico en metros.
        angulo_grados (float): Ángulo de articulación en grados sexagesimales.

    Retorna:
        posicion_x, posicion_y (float, float): Coordenadas espaciales resultantes.
    """
    # Conversión requerida para las funciones de la librería math
    angulo_rad = math.radians(angulo_grados)

    # Ecuaciones de transformación geométrica
    posicion_x = longitud_eslabon * math.cos(angulo_rad)
    posicion_y = longitud_eslabon * math.sin(angulo_rad)

    # Retorno múltiple
    return posicion_x, posicion_y


# Ejecución de la función
l1 = 0.5    # 0.5 metros
theta1 = 30.0 # 30 grados

# Desempaquetado directo de variables
x_efector, y_efector = resolver_cinematica_directa(l1, theta1)

print(f"Eslabón: L = {l1} m, θ = {theta1}°")
print(f"Coordenadas del efector final: X = {x_efector:.3f} m, Y = {y_efector:.3f} m")
```

## 2. Procesamiento Conciso de Arreglos (*List Comprehensions*)

El filtrado y la transformación de arreglos de datos provenientes de sensores (como cámaras, LiDAR o codificadores ópticos) son tareas continuas en robótica.

En C++, estas operaciones demandan la declaración de un nuevo vector y un bucle `for` iterativo. Python ofrece el paradigma de **Comprensión de Listas** (*List Comprehension*), permitiendo filtrar y transformar conjuntos de datos mediante una sintaxis compacta y de alto rendimiento.

```python
# ==============================================================================
# 2. FILTRADO Y TRANSFORMACIÓN DE SEÑALES SEN SÓRICAS
# ==============================================================================

print("\n--- 2. Filtrado de Datos Infrarrojos ---")

# Arreglo simulado con lecturas de distancia en cm (incluye ruido fuera de rango)
lecturas_raw_cm = [12.0, -5.0, 45.3, 120.0, 8.5, -1.0, 30.0]
rango_maximo_cm = 80.0

print(f"Lecturas crudas: {lecturas_raw_cm}")

# Comprensión de Listas para Filtrar datos inválidos (valores < 0 o > rango máximo)
# Sintaxis: [ expresión for elemento in lista if condición ]
lecturas_validas_cm = [d for d in lecturas_raw_cm if 0.0 <= d <= rango_maximo_cm]

print(f"Lecturas válidas (filtradas): {lecturas_validas_cm}")

# Comprensión de Listas para Transformación Matemática (Conversión cm -> metros)
lecturas_metros = [d / 100.0 for d in lecturas_validas_cm]

print(f"Lecturas convertidas a metros: {lecturas_metros}")
```

## 3. Programación Orientada a Objetos (POO) aplicada a la Robótica

En la mecatrónica moderna, la POO es el pilar sobre el cual se construyen marcos de trabajo como ROS (Robot Operating System). Una **Clase** funciona como un plano o plantilla para representar componentes del mundo real (sensores, actuadores, robots completos), encapsulando:
1. **Atributos:** El estado interno de la entidad (ej. posición, velocidad, estado de batería).
2. **Métodos:** Las acciones o comportamientos que la entidad puede ejecutar (ej. mover, detener, leer sensor).

*   **El método `__init__`:** Corresponde al **constructor** de la clase (equivalente al constructor en C++).
*   **El parámetro `self`:** Hace referencia a la instancia específica que ejecuta el código (equivalente al puntero `this` en C++).

```python
# ==============================================================================
# 3. ENCAPSULACIÓN: CLASES Y OBJETOS
# ==============================================================================

print("\n--- 3. Modelado de Hardware Orientado a Objetos ---")

# Definición de la clase base para un Motor de Corriente Continua
class MotorDC:

    # El constructor: Se ejecuta automáticamente al crear (instanciar) el objeto
    def __init__(self, id_motor, rpm_maximas):
        # Atributos (Estado del objeto). El uso de 'self.' los hace persistentes.
        self.id_motor = id_motor
        self.rpm_maximas = rpm_maximas
        self.rpm_actuales = 0.0
        self.estado = "Apagado"

    # Método (Comportamiento del objeto)
    def establecer_velocidad(self, porcentaje):
        """Asigna la velocidad del motor en función de un porcentaje (0 a 100)."""
        # Validación de seguridad algorítmica
        if porcentaje < 0: porcentaje = 0
        if porcentaje > 100: porcentaje = 100

        self.rpm_actuales = (porcentaje / 100.0) * self.rpm_maximas
        self.estado = "En movimiento" if porcentaje > 0 else "Detenido"

        print(f"[{self.id_motor}] Comandado al {porcentaje}%. RPM generadas: {self.rpm_actuales:.1f}")

    # Método para reportar la telemetría del motor
    def leer_telemetria(self):
        return f"Motor: {self.id_motor} | Estado: {self.estado} | Velocidad: {self.rpm_actuales} RPM"

# Instanciación: Creación de objetos físicos virtuales a partir de la clase
motor_izquierdo = MotorDC(id_motor="MTR-IZQ", rpm_maximas=3000)
motor_derecho = MotorDC(id_motor="MTR-DER", rpm_maximas=3000)

# Interacción con los objetos mediante sus métodos
motor_izquierdo.establecer_velocidad(50)  # Comando al 50%
motor_derecho.establecer_velocidad(100)   # Comando al 100%

print(f"\nReporte del sistema:")
print(motor_izquierdo.leer_telemetria())
print(motor_derecho.leer_telemetria())
```

## 4. Robustez Operacional: Manejo de Excepciones

En la interacción software-hardware, pueden surgir imponderables: pérdida de comunicación en un puerto serie, lectura de datos corruptos o indeterminaciones matemáticas (división por cero).

Un programa sin tolerancia a fallos colapsará (*crash*), dejando actuadores sin control y poniendo en riesgo la integridad física de la maquinaria. Los bloques `try`, `except`, `else` y `finally` permiten capturar excepciones en tiempo de ejecución y ejecutar rutinas de parada segura.

`else` y `finally` son opcionales. `finally` en particular tiene un comportamiento poco entendido por programadores.  Si desean saber más sobre su fucnionamiento y porque es importante les recomendamos el siguiente video
[Is "finally" Useless In Python?](https://www.youtube.com/watch?v=92JdbyISpCo)

```python
# ==============================================================================
# 4. TOLERANCIA A FALLOS Y SEGURIDAD ALGORÍTMICA
# ==============================================================================

print("\n--- 4. Control Seguro de Relación de Transmisión ---")

def calcular_torque_salida(torque_motor, relacion_reduccion):
    """
    Calcula el torque en el eje de salida.
    Protege el sistema ante relaciones de transmisión nulas o inválidas.
    """
    try:
        # Intento de cálculo sujeto a posibles errores matemáticos
        torque_salida = torque_motor * relacion_reduccion
        if relacion_reduccion <= 0:
            # Excepciones como ZeroDivisionError se generar automaticamente al ocurrir un error
            # Sin embargo, como programadores podemos generar (raise) excepciones si queremos
            # restringir algún funcionamiento u operació de forma adicional.
            # Como en este caso, que aunque matemáticamente tiene sentido dividir entre un valor
            # negativo, no tiene sentido físico aplicar una reducción negativa a un motor.
            raise ValueError("La relación de reducción debe ser estrictamente positiva.")

        relacion_inversa = 1.0 / relacion_reduccion

    # Podemos procesar cada excepción de forma independiente
    except ZeroDivisionError:
        print("¡EXCEPCIÓN INTERCEPTADA!: La relación de reducción no puede ser 0.")
        return None

    except ValueError as error_msg:
        print(f"¡EXCEPCIÓN INTERCEPTADA!: {error_msg}")
        return None

    else:
        # Se ejecuta ÚNICAMENTE si la sección 'try' finalizó sin ningún error
        print(f"Cálculo exitoso. Torque de salida: {torque_salida:.2f} Nm")
        return torque_salida

    finally:
        # Se ejecuta SIEMPRE. Ideal para liberar puertos de red o detener PWM por seguridad.
        print("Rutina de verificación finalizada (Estatus de hardware verificado).")

# Prueba 1: Ejecución con parámetros válidos
print("--- Caso 1: Parámetros Normales ---")
calcular_torque_salida(torque_motor=2.5, relacion_reduccion=5.0)

# Prueba 2: Fallo inducido por valor negativo
print("\n--- Caso 2: Error de Parámetro (Negativo) ---")
calcular_torque_salida(torque_motor=2.5, relacion_reduccion=-2.0)
```
