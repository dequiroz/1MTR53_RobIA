## Introducción

El presente cuaderno interactivo está diseñado para establecer los fundamentos del lenguaje Python, orientado específicamente a aplicaciones de Robótica e Inteligencia Artificial (IA).

Dado su conocimiento previo en lenguajes compilados como C y C++, notarán que Python prioriza la legibilidad y la rapidez en la creación de prototipos. Las diferencias principales radican en:
1. **Ausencia de llaves `{ }`:** Los bloques de código se definen estrictamente por la **indentación** (espacios al inicio de la línea).
2. **Tipado dinámico:** No es necesario declarar si una variable es `int`, `float` o `char`; el intérprete lo asigna automáticamente.
3. **Ausencia de punto y coma `;`:** Las instrucciones terminan con un salto de línea estándar.

```python
# ==============================================================================
# 1. SINTAXIS BÁSICA Y TIPADO DINÁMICO
# ==============================================================================

# En C/C++, la declaración exige el tipo: float voltaje_bateria = 12.5;
# En Python, la asignación es directa:

voltaje_bateria = 12.5      # Tipo flotante (float)
motores_activos = 4         # Tipo entero (int)
modo_autonomo = True        # Tipo booleano (bool). Nótese la primera letra en mayúscula.
identificador = "RBT-Alpha" # Tipo cadena de caracteres (str)

# La función type() permite identificar el tipo de dato asignado.
# Utilizamos "f-strings" (f"...") para concatenar texto y variables fácilmente,
# reemplazando la complejidad del printf() de C.

print("--- Inicialización del Sistema ---")
print(f"Robot: {identificador} | Voltaje: {voltaje_bateria}V")
print(f"El tipo de dato de 'motores_activos' es: {type(motores_activos)}")

# PRECAUCIÓN ALGORÍTMICA: Al no tener un tipo fijo, una variable puede cambiar
# su naturaleza durante la ejecución. Esto debe evitarse en sistemas de control
# para mantener la estabilidad del código.
```

## 2. Álgebra de Boole y Toma de Decisiones

El diseño de algoritmos reactivos requiere la evaluación constante del entorno mediante sensores. En C, los operadores lógicos son matemáticos (`&&`, `||`, `!`). Python utiliza un enfoque de lenguaje natural (`and`, `or`, `not`), lo que facilita la transcripción directa desde un diagrama de flujo o seudocódigo lógico.

```python
# ==============================================================================
# 2. OPERADORES LÓGICOS Y CONDICIONALES (IF - ELIF - ELSE)
# ==============================================================================

# Variables simuladas a partir de la lectura de un sensor ultrasónico (en cm)
distancia_frontal = 12.0
distancia_lateral = 45.0
bateria_critica = False

print("\n--- Subsistema de Navegación y Evasión ---")

# Evaluación lógica combinada.
# La indentación define qué código se ejecuta si se cumple la condición.

if distancia_frontal < 15.0 and distancia_lateral < 15.0:
    # Ambas condiciones deben cumplirse (AND)
    print("ALERTA: Entorno cerrado. Comandos: Detener motores y emitir alarma.")
    velocidad = 0.0

elif distancia_frontal < 15.0 or distancia_lateral < 15.0:
    # Al menos una condición se cumple (OR) - 'elif' equivale a 'else if' en C
    print("PRECAUCIÓN: Obstáculo detectado en un eje. Comandos: Reducir velocidad al 30%.")
    velocidad = 0.3

else:
    # Ninguna de las anteriores se cumplió
    print("VÍA LIBRE: Sin obstáculos cercanos. Comandos: Velocidad nominal.")
    velocidad = 1.0

# Uso del operador de negación (NOT) equivalente a '!' en C
if not bateria_critica:
    print(f"Estado de energía óptimo. Ejecutando motores a {velocidad * 100}% de potencia.")
```

## 3. Construcción de Algoritmos Iterativos (Bucles)

Los lazos de control cerrados y el procesamiento de señales dependen de iteraciones constantes. Python simplifica los bucles clásicos:
*   **Bucle `while`:** Ideal para procesos que deben ejecutarse continuamente hasta que una condición física o lógica cambie.
*   **Bucle `for`:** A diferencia de C, donde se requiere un contador explícito (`for(int i=0; i<N; i++)`), en Python el `for` itera directamente sobre secuencias o rangos predefinidos.

```python
# ==============================================================================
# 3. PATRONES ITERATIVOS: WHILE Y FOR
# ==============================================================================

print("\n--- Secuencia de Calibración (Uso de 'while') ---")

# Patrón algorítmico: Bucle con condición de salida
presion_sistema = 0.0
presion_deseada = 2.5
ciclo = 0

# Se ejecuta mientras la presión sea menor a la deseada
while presion_sistema < presion_deseada:
    ciclo += 1                # Operador de incremento (equivale a ciclo = ciclo + 1)
    presion_sistema += 0.75   # Simulamos la acción de una bomba
    print(f"Ciclo {ciclo} -> Presión actual: {presion_sistema:.2f} atm")

print("Calibración del sistema hidráulico completada.")


print("\n--- Barrido de Actuador (Uso de 'for') ---")

# La función range(inicio, fin, paso) genera una secuencia automática.
# El 'fin' es un límite excluyente (llega hasta fin - 1).
# Este bucle va de 0 a 40 grados, con incrementos de 10.

for angulo in range(0, 50, 10):
    print(f"Enviando señal PWM: Posicionando servomotor en {angulo}°")
```

## 4. Estructuras de Datos, Indexación y Procesamiento

En mecatrónica es ineficiente crear variables individuales para múltiples datos (`sensor1`, `sensor2`, `sensor3`). Para organizar la información, Python ofrece tres estructuras fundamentales:

1.  **Listas (`[ ]`):** Arreglos dinámicos y mutables. Fundamentales para almacenar un flujo temporal de datos (ej. histórico de sensores).
2.  **Tuplas (`( )`):** Colecciones inmutables. Se usan para datos que no deben alterarse accidentalmente, como coordenadas $(x, y, z)$.
3.  **Diccionarios (`{ }`):** Estructuras de clave-valor. Ideales para mapear el estado completo de un sistema.

**Nota sobre indexación:** Al igual que en C, el primer elemento de una estructura siempre ocupa la **posición 0**.

```python
# ==============================================================================
# 4. ESTRUCTURAS DE DATOS Y PATRONES ALGORÍTMICOS
# ==============================================================================

print("\n--- 4.1 Listas e Indexación ---")
# Arreglo con lecturas consecutivas de un sensor LiDAR (en metros)
lecturas_lidar = [1.2, 1.5, 2.8, 3.1, 1.4]
print(f"Matriz de datos LiDAR: {lecturas_lidar}")

# Acceso al primer elemento (Índice 0)
print(f"Primera lectura registrada: {lecturas_lidar[0]} m")

# Slicing (Particionamiento): Extraer una sub-matriz de datos
# Extrae desde el índice 1 hasta el 3 (el límite superior 4 es excluyente)
segmento_central = lecturas_lidar[1:4]
print(f"Sub-segmento de datos (índices 1, 2, 3): {segmento_central}")

# Patrón algorítmico: Acumulador
# Recorremos la lista para aplicar un filtro de promedio simple
suma_lecturas = 0.0
for valor in lecturas_lidar:
    suma_lecturas += valor

promedio = suma_lecturas / len(lecturas_lidar) # len() obtiene la longitud del arreglo
print(f"El promedio de las mediciones es: {promedio:.2f} m")


print("\n--- 4.2 Tuplas (Estructuras Inmutables) ---")
# Representan estados físicos invariables en una iteración, como el Origen.
origen_cartesiano = (0.0, 0.0, 0.0)
print(f"Coordenadas de la base del robot: {origen_cartesiano}")
# origen_cartesiano[0] = 1.0  # Descomentar esto causaría un error, la tupla protege el dato.


print("\n--- 4.3 Diccionarios (Mapeo de Estados) ---")
# Agrupan atributos heterogéneos de forma semántica, similar a un 'struct' en C.
telemetria = {
    "rpm_motor1": 1200,
    "rpm_motor2": 1195,
    "temperatura_chasis": 35.5,
    "estado_comunicacion": "OK"
}

print("Datos de telemetría extraídos del diccionario:")
print(f"Velocidad Motor 1: {telemetria['rpm_motor1']} RPM")
print(f"Temperatura actual: {telemetria['temperatura_chasis']} °C")
```
